```python
import torch
import torch.nn as nn
```

```python
class LinearRegressionScratch( nn.Module ):
  def __init__(self,num_inputs) -> None:
    super().__init__()
    self.linear = nn.Linear(num_inputs , 1)

  def forward(self , X):
    return self.linear(X)
```
